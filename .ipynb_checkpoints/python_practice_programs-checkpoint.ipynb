### 3. Area of a Circle

```python
import math

r = float(input("Enter radius: "))
area = math.pi * r**2
print(f"Area of circle: {area:.2f}")
```

### 4. Factorial and Fibonacci Series

```python
import math

n = int(input("Enter a number: "))

# Factorial
print(f"Factorial of {n}: {math.factorial(n)}")

# Fibonacci Series up to n terms
a, b = 0, 1
fib = []
for _ in range(n):
    fib.append(a)
    a, b = b, a + b
print(f"Fibonacci series ({n} terms): {fib}")
```

### 5. String Length, First and Last Character

```python
s = input("Enter a string: ")
print(f"Length: {len(s)}")
if s:
    print(f"First character: {s[0]}")
    print(f"Last character: {s[-1]}")
```

### 6. Join Two Strings

```python
str1 = input("Enter first string: ")
str2 = input("Enter second string: ")
result = str1 + " " + str2
print("Joined string:", result)
```

### 7. Reverse a String

```python
s = input("Enter a string: ")
print("Reversed string:", s[::-1])
```

### 8. Count Vowels in a String

```python
s = input("Enter a string: ")
vowels = "aeiouAEIOU"
count = sum(1 for ch in s if ch in vowels)
print(f"Number of vowels: {count}")
```

### 9. Check Positive, Negative, or Zero

```python
num = float(input("Enter a number: "))
if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

### 13. Print 1 to N Using a For Loop

### 16. Sum of Numbers from 1 to N

```python
n = int(input("Enter n: "))
total = sum(range(1, n + 1))
print(f"Sum from 1 to {n}: {total}")
```

### 17. Factorial Using a For Loop

```python
n = int(input("Enter a number: "))
fact = 1
for i in range(1, n + 1):
    fact *= i
print(f"Factorial of {n}: {fact}")
```

### 18. Numbers from 10 to 1 Using a While Loop

```python
i = 10
while i >= 1:
    print(i, end=" ")
    i -= 1
print()
```

### 19. Skip Number 5 Using continue

```python
for i in range(1, 11):
    if i == 5:
        continue
    print(i, end=" ")
print()
```

### 20. Stop at 7 Using break

```python
for i in range(1, 11):
    if i == 7:
        break
    print(i, end=" ")
print()
```

### 21. Square Root and Power Using math

```python
import math

num = float(input("Enter number: "))
exp = float(input("Enter power/exponent: "))

print(f"Square root: {math.sqrt(num)}")
print(f"{num} raised to power {exp}: {math.pow(num, exp)}")
```

### 22. Custom Function: Sum of Two Numbers

```python
def add(a, b):
    return a + b

x = float(input("Enter first number: "))
y = float(input("Enter second number: "))
print(f"Sum: {add(x, y)}")
```

### 23. Custom Function: Area of Rectangle

```python
def rectangle_area(length, width):
    return length * width

l = float(input("Enter length: "))
w = float(input("Enter width: "))
print(f"Area of rectangle: {rectangle_area(l, w)}")
```

### 24. Handle Division by Zero Using try-except

```python
try:
    a = float(input("Enter numerator: "))
    b = float(input("Enter denominator: "))
    print(f"Result: {a / b}")
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
```

### 25. Handle Invalid Integer Input Using try-except

```python
try:
    val = int(input("Enter an integer: "))
    print(f"Valid integer entered: {val}")
except ValueError:
    print("Error: Invalid input! Please enter a valid whole number.")
```
