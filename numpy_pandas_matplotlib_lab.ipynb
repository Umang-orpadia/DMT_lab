## Section A: NumPy Programs

### 1. 4×4 Matrix, Diagonal Elements, and Trace

```python
import numpy as np

# Create a 4x4 matrix
matrix_4x4 = np.array([
    [10, 15, 20, 25],
    [30, 35, 40, 45],
    [50, 55, 60, 65],
    [70, 75, 80, 85]
])

# Extract diagonal elements and calculate trace
diagonal_elements = np.diag(matrix_4x4)
matrix_trace = np.trace(matrix_4x4)

print("4x4 Matrix:\n", matrix_4x4)
print("\nDiagonal Elements:", diagonal_elements)
print("Trace of the Matrix:", matrix_trace)
```

### 2. Concatenate Two Arrays

```python
import numpy as np

array1 = np.array([1, 2, 3, 4])
array2 = np.array([5, 6, 7, 8])

# Concatenate arrays
concatenated_array = np.concatenate((array1, array2))

print("Array 1:", array1)
print("Array 2:", array2)
print("Concatenated Array:", concatenated_array)
```

### 3. Replace Elements Greater than 50 with 100

```python
import numpy as np

arr = np.array([12, 55, 34, 78, 90, 45, 51, 23])
print("Original Array:", arr)

# Replace elements > 50 with 100
modified_arr = np.where(arr > 50, 100, arr)
print("Modified Array:", modified_arr)
```

### 4. Median, Variance, and Standard Deviation

```python
import numpy as np

data = np.array([14, 22, 19, 30, 45, 28, 35, 40])

median_val = np.median(data)
variance_val = np.var(data)
std_dev_val = np.std(data)

print("Dataset:", data)
print(f"Median: {median_val}")
print(f"Variance: {variance_val:.2f}")
print(f"Standard Deviation: {std_dev_val:.2f}")
```

### 5. Unique Elements and Their Frequencies

```python
import numpy as np

sample_array = np.array([10, 20, 10, 30, 20, 40, 10, 50, 20, 30])

unique_elements, counts = np.unique(sample_array, return_counts=True)

print("Array:", sample_array)
print("\nUnique Elements and Frequencies:")
for elem, count in zip(unique_elements, counts):
    print(f"Element {elem}: occurs {count} time(s)")
```

## Section B: Pandas Programs

### 6. Add a New Row to an Existing Student DataFrame

```python
import pandas as pd

# Initial Student DataFrame
students_df = pd.DataFrame({
    'StudentID': [101, 102, 103],
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Marks': [85, 78, 92]
})
print("Initial DataFrame:\n", students_df)

# Adding a new row
new_student = pd.DataFrame([{'StudentID': 104, 'Name': 'Diana', 'Marks': 88}])
students_df = pd.concat([students_df, new_student], ignore_index=True)

print("\nDataFrame after Adding Row:\n", students_df)
```

### 7. Identify and Display Missing Values in a DataFrame

```python
import pandas as pd
import numpy as np

df_missing = pd.DataFrame({
    'StudentID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', np.nan, 'Diana'],
    'Department': ['CS', 'EE', 'ME', np.nan],
    'Marks': [85, np.nan, 76, 91]
})

print("DataFrame:\n", df_missing)
print("\nBoolean Missing Value Mask:\n", df_missing.isnull())
print("\nTotal Missing Values Per Column:\n", df_missing.isnull().sum())
```

### 8. Remove Duplicate Rows from a DataFrame

```python
import pandas as pd

data_duplicates = pd.DataFrame({
    'StudentID': [101, 102, 101, 103, 102],
    'Name': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob'],
    'Marks': [85, 78, 85, 92, 78]
})
print("Before Dropping Duplicates:\n", data_duplicates)

# Dropping duplicate rows
df_dedup = data_duplicates.drop_duplicates()
print("\nAfter Dropping Duplicates:\n", df_dedup)
```

### 9. Group Records by Department and Calculate Average Marks

```python
import pandas as pd

records = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Department': ['CS', 'EE', 'CS', 'ME', 'EE', 'ME'],
    'Marks': [85, 76, 92, 80, 84, 88]
})

avg_marks_by_dept = records.groupby('Department')['Marks'].mean().reset_index()
avg_marks_by_dept.rename(columns={'Marks': 'AverageMarks'}, inplace=True)

print("Original Records:\n", records)
print("\nAverage Marks by Department:\n", avg_marks_by_dept)
```

### 10. Assign Grades Based on Marks using `apply()`

```python
import pandas as pd

students = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Marks': [94, 82, 73, 58, 65]
})

def assign_grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    else:
        return 'D'

students['Grade'] = students['Marks'].apply(assign_grade)
print("Students with Assigned Grades:\n", students)
```

## Section C: Matplotlib Visualization Programs

### 11. Line Chart: Sales of a Product Over Five Months

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [250, 310, 280, 420, 390]

plt.figure(figsize=(7, 4))
plt.plot(months, sales, marker='o', color='royalblue', linewidth=2, linestyle='-')
plt.title('Monthly Sales Trend (5 Months)', fontsize=13, pad=12)
plt.xlabel('Month', fontsize=11)
plt.ylabel('Units Sold', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### 12. Bar Chart: Comparing Marks of Five Students

```python
import matplotlib.pyplot as plt

students = ['Alice', 'Bob', 'Charlie', 'Diana', 'Evan']
marks = [85, 92, 78, 88, 95]

plt.figure(figsize=(7, 4))
bars = plt.bar(students, marks, color='teal', width=0.55)
plt.title('Comparison of Student Marks', fontsize=13, pad=12)
plt.xlabel('Student Name', fontsize=11)
plt.ylabel('Marks', fontsize=11)
plt.ylim(0, 105)

# Value labels above bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, yval + 1.5, f'{yval}', ha='center', va='bottom')

plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### 13. Pie Chart: Percentage Distribution of Students Across Departments

```python
import matplotlib.pyplot as plt

departments = ['Computer Science', 'Mechanical', 'Electrical', 'Civil', 'Electronics']
student_counts = [120, 80, 65, 45, 90]
colors = ['#4e79a7', '#f28e2b', '#e15759', '#76b7b2', '#59a14f']

plt.figure(figsize=(6, 6))
plt.pie(
    student_counts,
    labels=departments,
    autopct='%1.1f%%',
    startangle=140,
    colors=colors,
    explode=[0.05, 0, 0, 0, 0]
)
plt.title('Student Distribution Across Departments', fontsize=13, pad=12)
plt.show()
```

### 14. Scatter Plot: Study Hours vs Examination Marks

```python
import matplotlib.pyplot as plt

study_hours = [1.5, 2.0, 3.2, 4.0, 4.5, 5.0, 6.2, 7.0, 8.1, 9.0]
exam_marks = [45, 52, 60, 68, 70, 78, 85, 88, 93, 97]

plt.figure(figsize=(7, 4))
plt.scatter(study_hours, exam_marks, color='crimson', edgecolors='black', s=70)
plt.title('Study Hours vs. Examination Marks', fontsize=13, pad=12)
plt.xlabel('Study Hours (per day)', fontsize=11)
plt.ylabel('Examination Marks', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
