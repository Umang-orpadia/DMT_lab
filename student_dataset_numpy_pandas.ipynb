### 1. Load Dataset and Inspect Shape & Columns (Q10)
Reads `student-dataset.csv` and displays its data, column names, number of rows, and number of columns.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('student-dataset.csv')

print(f"Total Rows: {df.shape[0]}")
print(f"Total Columns: {df.shape[1]}")
print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
display(df.head())
```

### 2. Display First 3 and Last 3 Records (Q8)
Using `.head(3)` and `.tail(3)`.

```python
print("--- First 3 Records ---")
display(df[['id', 'name', 'nationality', 'math.grade', 'sciences.grade']].head(3))

print("--- Last 3 Records ---")
display(df[['id', 'name', 'nationality', 'math.grade', 'sciences.grade']].tail(3))
```

### 3. Extract a Pandas Series of Student Grades (Q6)
Creating and displaying a Series of student Math grades with student names as the index.

```python
math_series = pd.Series(df['math.grade'].values[:10], index=df['name'][:10], name="Math_Grade")
print("Top 10 Students Math Grades Series:")
print(math_series)
```

### 4. Create a Sub-DataFrame with Name, ID, and Grades (Q7)
Selecting relevant student identifiers and their academic marks.

```python
student_summary_df = df[['id', 'name', 'gender', 'age', 'math.grade', 'sciences.grade', 'english.grade']].copy()
print("Subset DataFrame:")
display(student_summary_df.head())
```

### 5. Calculate Total and Average Marks across Subjects (Q9)
Computing total grade across subjects and class-wide averages.

```python
subjects = ['english.grade', 'math.grade', 'sciences.grade', 'language.grade']

df['Total_Marks'] = df[subjects].sum(axis=1)
df['Average_Marks'] = df[subjects].mean(axis=1)

print(f"Overall Average Math Grade: {df['math.grade'].mean():.2f}")
print(f"Overall Average Total Grade: {df['Total_Marks'].mean():.2f}")
display(df[['name', 'Total_Marks', 'Average_Marks']].head())
```

### 6. Filter High-Scoring Students (Q11)
Filtering students with an average grade greater than 3.5 (or Math grade > 3.8).

```python
top_math_students = df[df['math.grade'] >= 3.8][['id', 'name', 'nationality', 'math.grade', 'sciences.grade']]
print(f"Number of students with Math Grade >= 3.8: {len(top_math_students)}")
display(top_math_students.head(10))
```

### 7. Find Highest and Lowest Marks (Q12)
Using Pandas `.max()` and `.min()` and finding the corresponding student records.

```python
print("Math Grade Stats:")
print(f"  Highest: {df['math.grade'].max()}")
print(f"  Lowest:  {df['math.grade'].min()}")

top_student = df.loc[df['math.grade'].idxmax()]
lowest_student = df.loc[df['math.grade'].idxmin()]

print(f"\nTop Math Student: {top_student['name']} ({top_student['math.grade']})")
print(f"Lowest Math Student: {lowest_student['name']} ({lowest_student['math.grade']})")
```

### 8. Sort Students by Marks (Ascending & Descending) (Q13)
Sorting the DataFrame based on academic performance.

```python
print("Top 5 Ranked Students (Descending by Total Marks):")
display(df[['name', 'nationality', 'Total_Marks']].sort_values(by='Total_Marks', ascending=False).head())

print("\nBottom 5 Ranked Students (Ascending by Total Marks):")
display(df[['name', 'nationality', 'Total_Marks']].sort_values(by='Total_Marks', ascending=True).head())
```

### 9. Add 'Result' Column Based on Threshold (Q14)
Adding a conditional `Result` column ('Pass' if average >= 2.5 else 'Fail').

```python
df['Result'] = df['Average_Marks'].apply(lambda x: 'Pass' if x >= 2.5 else 'Fail')

print("Result Distribution:")
print(df['Result'].value_counts())

display(df[['name', 'Average_Marks', 'Result']].head(10))
```

### 10. NumPy Integration with Dataset (NumPy Q1 - Q5)
Using NumPy arrays directly extracted from the dataset columns.

```python
math_array = df['math.grade'].to_numpy()
print("NumPy 1D Array (first 10 elements):", math_array[:10])

print(f"Sum: {np.sum(math_array):.2f}")
print(f"Mean: {np.mean(math_array):.2f}")
print(f"Max: {np.max(math_array):.2f}")
print(f"Min: {np.min(math_array):.2f}")

grade_matrix = df[['math.grade', 'sciences.grade', 'english.grade']].to_numpy()
print("\n2D Grade Matrix Shape (Rows, Columns):", grade_matrix.shape)

ages = df['age'].to_numpy()
even_ages = ages[ages % 2 == 0]
print(f"Total students with even age: {len(even_ages)}")
```
